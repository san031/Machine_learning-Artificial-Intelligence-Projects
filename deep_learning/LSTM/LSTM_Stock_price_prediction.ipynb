# Part 1: Data preprocessing

Step 1: Importing libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

#### step 2: importing training set

```python
dataset_train = pd.read_csv('/content/sample_data/Google_Stock_Price_Train.csv')
```

```python
dataset_train.head()
```

```python
training_set = dataset_train.iloc[:,1:2].values
```

```python
training_set
```

#### step 3: feature Scaling

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
sc = MinMaxScaler(feature_range=(0,1))
training_set_scaled = sc.fit_transform(training_set)
```

```python
training_set_scaled
```

#### step 4 : creating a datastructure with 60 timestamps

```python
X_train = []
y_train = []
for i in range(60,1258):
  X_train.append(training_set_scaled[i-60:i,0])    #59th day
  y_train.append(training_set_scaled[i,0])         #60th day output value

X_train,y_train = np.array(X_train),np.array(y_train)
```

```python
X_train
```

#### step5 : Reshaping

```python
X_train = np.reshape(X_train,(X_train.shape[0],X_train.shape[1],1))
```

```python
X_train.shape
```

# Part 2 : Building and Training the RNN

#### Step 1 : Building and importing libraries and packages

```python
from keras.models import Sequential
from keras.layers import Dense,LSTM,Dropout
```

```python
regressor = Sequential()
```

#### Adding the first LSTM Layers and some Dropout regularisation

```python
regressor.add(LSTM(units = 50, return_sequences = True, input_shape = (X_train.shape[1],1)))
regressor.add(Dropout(0.2))
```

#### adding second layer (LSTM) and dropout

```python
regressor.add(LSTM(units = 50, return_sequences=True))
regressor.add(Dropout(0.2))
```

#### adding third layer (LSTM) and dropout



```python
regressor.add(LSTM(units = 50, return_sequences=True))
regressor.add(Dropout(0.2))
```

#### adding output layer

```python
regressor.add(Dense(units = 1))
```

#### compiling the RNN

```python
regressor.compile(optimizer='adam',loss = 'mean_squared_error')
```

#### model training  

```python
regressor.fit(X_train,y_train,epochs = 100,batch_size = 32)
```

```python
regressor.save("stock_model.keras")
```
