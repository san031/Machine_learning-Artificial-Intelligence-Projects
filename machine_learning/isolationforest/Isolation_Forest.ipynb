```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/sample_data/tr_server_data.csv",header = None, names = ["feature_1","feature_2"])
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(df)
```

### train model using isolation forest


```python
from sklearn.ensemble import IsolationForest
```

```python
model = IsolationForest(
    n_estimators=200,
    contamination=0.05,
    random_state=42
)
```

```python
model.fit(X_scaled)
```

### detect anamolies

```python
## prediction
## -1 = anamoly, 1 = normal
```

```python
df["anamoly"] = model.predict(X_scaled)
```
